# Regime Analysis & Backtest

Explore distributions, transitions, and strategy performance.

```python
import pandas as pd
import numpy as np
from signals import compute_signals

results = pd.read_csv('regime_results.csv', index_col=0, parse_dates=True)
print(f'Loaded {len(results)} trading days')
```

## Regime Distributions

```python
signals = compute_signals(results)
for regime, stats in signals['distributions'].items():
    print(f'{regime}: vol={stats["ann_vol"]:.1%}')
```

## Current Regime Context

```python
print(f'Current: {signals["current_regime"]}')
print(f'Confidence: {signals["awareness"]["confidence"]:.1%}')
```

## Regime Duration Statistics

```python
regimes = results['regime_label'].values
print(f'Total periods: {len(regimes)}')
for r in set(regimes):
    pct = (regimes == r).sum() / len(regimes)
    print(f'{r}: {pct:.1%}')
```

## Position Sizing Strategy

```python
kelly_adj = {'Low-Vol': 1.5, 'Med-Vol': 1.0, 'High-Vol': 0.5}
position_sizes = [0.25 * kelly_adj.get(r, 1.0) for r in regimes]
print(f'Position sizing applied to {len(position_sizes)} periods')
```
